```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv(r"C:\Users\sneha\Downloads\movie_data_merged.csv")
```

```python
df
```

```python
df = df.dropna(subset=["description", "genre"])
df["description"] = df["description"].astype(str)

x=df["description"]
y=df["genre"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(df["description"], df["genre"], test_size=0.2, random_state=42)
```

```python
vectorizer = TfidfVectorizer(stop_words="english", max_features=5000)
x_train = vectorizer.fit_transform(x_train)
x_test = vectorizer.transform(x_test)
```

```python
model = LogisticRegression(max_iter=200, solver="liblinear")
model.fit(x_train, y_train)

```

```python
pred = model.predict(x_test)
print("Accuracy:", accuracy_score(y_test, pred))
```

```python
new_movie = ["A detective investigates a mysterious murder."]
print("Genre:", model.predict(
    vectorizer.transform(new_movie)
)[0])
```
